```python
import pandas as pd
df = pd.read_csv('Marketing.csv')
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%d/%m/%Y',
    errors='coerce'
)
df.dropna(subset=['Date'], inplace=True)
df.set_index('Date', inplace=True)
df.sort_index(inplace=True)
sliced_data = df.loc['2020-11-11':'2020-11-15']
print(sliced_data)
```
